#Prepare classification crops


```python
from pathlib import Path
from PIL import Image
from collections import Counter

root = Path(extract_dir)
crop_root = Path("/content/plantdoc_classification")

class_names = data_config["names"]
image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
counts = Counter()
skipped = Counter()

for split in ["train", "valid", "test"]:
    split_dir = root / split
    images_dir = split_dir / "images"
    labels_dir = split_dir / "labels"

    label_map = {
        p.relative_to(labels_dir).with_suffix("").as_posix(): p
        for p in labels_dir.rglob("*.txt")
    }

    for img_path in images_dir.rglob("*"):
        if img_path.suffix.lower() not in image_exts:
            continue

        key = img_path.relative_to(images_dir).with_suffix("").as_posix()
        label_path = label_map.get(key)

        if label_path is None:
            skipped["missing_label"] += 1
            continue

        with open(label_path, "r") as f:
            lines = [line.split() for line in f if line.strip()]

        if not lines:
            skipped["empty_label"] += 1
            continue

        try:
            annotations = [
                (int(row[0]), list(map(float, row[1:])))
                for row in lines if len(row) == 5
            ]
        except ValueError:
            skipped["invalid_label"] += 1
            continue

        if not annotations:
            skipped["invalid_label"] += 1
            continue

        class_ids = {a[0] for a in annotations}

        if len(class_ids) != 1:
            skipped["multiple_classes"] += 1
            continue

        class_id = next(iter(class_ids))

        if not 0 <= class_id < len(class_names):
            skipped["unknown_class"] += 1
            continue

        try:
            with Image.open(img_path) as im:
                im = im.convert("RGB")
                w, h = im.size

                for box_id, (_, box) in enumerate(annotations):
                    xc, yc, bw, bh = box

                    if not all(0 <= v <= 1 for v in box):
                        skipped["invalid_box"] += 1
                        continue

                    x1 = max(0, int((xc - bw / 2) * w))
                    y1 = max(0, int((yc - bh / 2) * h))
                    x2 = min(w, int((xc + bw / 2) * w))
                    y2 = min(h, int((yc + bh / 2) * h))

                    if x2 <= x1 or y2 <= y1:
                        skipped["invalid_box"] += 1
                        continue

                    crop = im.crop((x1, y1, x2, y2))

                    if crop.width < 16 or crop.height < 16:
                        skipped["small_crop"] += 1
                        continue

                    class_folder = crop_root / split / str(class_id)
                    class_folder.mkdir(parents=True, exist_ok=True)

                    out_path = class_folder / f"{img_path.stem}_{box_id}.jpg"
                    crop.save(out_path, quality=95)
                    counts[(split, class_id)] += 1

        except Exception:
            skipped["image_error"] += 1

print("Crop preparation complete.")
print("Crops per split:", {
    split: sum(v for (s, _), v in counts.items() if s == split)
    for split in ["train", "valid", "test"]
})
print("Skipped:", dict(skipped))
print("Output:", crop_root)
```

```python
from pathlib import Path
import shutil

source_root = Path("/content/plantdoc_classification")
filtered_root = Path("/content/plantdoc_classification_filtered")

splits = ["train", "valid", "test"]

# Find classes that contain images in every split
class_sets = []

for split in splits:
    split_dir = source_root / split
    available = {
        folder.name
        for folder in split_dir.iterdir()
        if folder.is_dir() and any(
            file.suffix.lower() in {".jpg", ".jpeg", ".png", ".webp"}
            for file in folder.iterdir()
        )
    }
    class_sets.append(available)

common_classes = sorted(
    set.intersection(*class_sets),
    key=int
)

print("Common classes:", common_classes)
print("Number of classes:", len(common_classes))

# Create the filtered dataset
if filtered_root.exists():
    shutil.rmtree(filtered_root)

for split in splits:
    for class_id in common_classes:
        src = source_root / split / class_id
        dst = filtered_root / split / class_id
        dst.mkdir(parents=True, exist_ok=True)

        for image_path in src.iterdir():
            if image_path.suffix.lower() in {
                ".jpg", ".jpeg", ".png", ".webp"
            }:
                shutil.copy2(image_path, dst / image_path.name)

print("Filtered dataset created:", filtered_root)
```

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from pathlib import Path

filtered_root = Path("/content/plantdoc_classification_filtered")

image_size = 224
batch_size = 32

# ImageNet normalization
mean = (0.485, 0.456, 0.406)
std = (0.229, 0.224, 0.225)

# Training transformations
train_transform = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# Validation and test transformations
eval_transform = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# Load datasets
train_dataset = datasets.ImageFolder(
    str(filtered_root / "train"),
    transform=train_transform
)

valid_dataset = datasets.ImageFolder(
    str(filtered_root / "valid"),
    transform=eval_transform
)

test_dataset = datasets.ImageFolder(
    str(filtered_root / "test"),
    transform=eval_transform
)

# Confirm that all splits use the same class mapping
assert train_dataset.class_to_idx == valid_dataset.class_to_idx
assert train_dataset.class_to_idx == test_dataset.class_to_idx

# Create data loaders
train_loader = DataLoader(
    train_dataset, batch_size=batch_size, shuffle=True,
    num_workers=2, pin_memory=True
)

valid_loader = DataLoader(
    valid_dataset, batch_size=batch_size, shuffle=False,
    num_workers=2, pin_memory=True
)

test_loader = DataLoader(
    test_dataset, batch_size=batch_size, shuffle=False,
    num_workers=2, pin_memory=True
)

print("Data loaders created successfully.")
print("Number of classes:", len(train_dataset.classes))
print("Training images:", len(train_dataset))
print("Validation images:", len(valid_dataset))
print("Testing images:", len(test_dataset))
print("Class mapping:", train_dataset.class_to_idx)
print("Device:", "CUDA GPU" if torch.cuda.is_available() else "CPU")
```

# Build the CNN baseline


```python
import torch
import torch.nn as nn
import torch.optim as optim

# Use the available GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Define a simple CNN
class PlantDiseaseCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)

# Create model, loss function, and optimizer
cnn_model = PlantDiseaseCNN(
    num_classes=len(train_dataset.classes)
).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(cnn_model.parameters(), lr=0.001)

print("CNN model created successfully.")
print("Device:", device)
print("Number of classes:", len(train_dataset.classes))
print("Trainable parameters:",
      sum(p.numel() for p in cnn_model.parameters() if p.requires_grad))
```

##Train the CNN

```python
import time
import torch

num_epochs = 10
best_val_accuracy = 0.0
cnn_history = []

for epoch in range(num_epochs):
    start_time = time.time()

    # Training phase
    cnn_model.train()
    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = cnn_model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item() * images.size(0)
        predictions = outputs.argmax(dim=1)
        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

    train_loss /= train_total
    train_accuracy = train_correct / train_total

    # Validation phase
    cnn_model.eval()
    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in valid_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = cnn_model(images)
            loss = criterion(outputs, labels)

            val_loss += loss.item() * images.size(0)
            predictions = outputs.argmax(dim=1)
            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    val_loss /= val_total
    val_accuracy = val_correct / val_total

    cnn_history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_accuracy
    })

    if val_accuracy > best_val_accuracy:
        best_val_accuracy = val_accuracy
        torch.save(cnn_model.state_dict(), "/content/best_cnn_model.pth")

    elapsed = time.time() - start_time

    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Time: {elapsed:.1f}s"
    )

print(f"\nBest validation accuracy: {best_val_accuracy:.4f}")
print("Best model saved to /content/best_cnn_model.pth")
```

## Build the Vision Transformer (ViT)

```python
import torch
import torch.nn as nn
from torchvision.models import vit_b_16, ViT_B_16_Weights

# Load pretrained Vision Transformer
vit_model = vit_b_16(weights=ViT_B_16_Weights.DEFAULT)

# Replace the final layer for our 27 classes
num_features = vit_model.heads.head.in_features
vit_model.heads.head = nn.Linear(
    num_features,
    len(train_dataset.classes)
)

# Move model to GPU
vit_model = vit_model.to(device)

# Loss function and optimizer
vit_criterion = nn.CrossEntropyLoss()
vit_optimizer = torch.optim.AdamW(
    vit_model.parameters(),
    lr=0.0001,
    weight_decay=0.01
)

print("Vision Transformer created successfully.")
print("Device:", device)
print("Number of classes:", len(train_dataset.classes))
print(
    "Trainable parameters:",
    sum(p.numel() for p in vit_model.parameters() if p.requires_grad)
)
```

#Train the Vision Transformer

```python
import time
import torch

num_epochs = 5
best_vit_val_accuracy = 0.0
vit_history = []

for epoch in range(num_epochs):
    start_time = time.time()

    # Training phase
    vit_model.train()
    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)

        vit_optimizer.zero_grad()
        outputs = vit_model(images)
        loss = vit_criterion(outputs, labels)
        loss.backward()
        vit_optimizer.step()

        train_loss += loss.item() * images.size(0)
        train_correct += (outputs.argmax(dim=1) == labels).sum().item()
        train_total += labels.size(0)

    train_loss /= train_total
    train_accuracy = train_correct / train_total

    # Validation phase
    vit_model.eval()
    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in valid_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = vit_model(images)
            loss = vit_criterion(outputs, labels)

            val_loss += loss.item() * images.size(0)
            val_correct += (outputs.argmax(dim=1) == labels).sum().item()
            val_total += labels.size(0)

    val_loss /= val_total
    val_accuracy = val_correct / val_total

    vit_history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_accuracy
    })

    if val_accuracy > best_vit_val_accuracy:
        best_vit_val_accuracy = val_accuracy
        torch.save(vit_model.state_dict(), "/content/best_vit_model.pth")

    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Time: {time.time() - start_time:.1f}s"
    )

print(f"\nBest ViT validation accuracy: {best_vit_val_accuracy:.4f}")
print("Best model saved to /content/best_vit_model.pth")
```

# data-augmentation experiment
## ViT trained without random augmentation.

```python
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Training images without random augmentation
no_aug_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=(0.485, 0.456, 0.406),
        std=(0.229, 0.224, 0.225)
    )
])

no_aug_train_dataset = datasets.ImageFolder(
    "/content/plantdoc_classification_filtered/train",
    transform=no_aug_transform
)

no_aug_train_loader = DataLoader(
    no_aug_train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

assert no_aug_train_dataset.class_to_idx == valid_dataset.class_to_idx

print("Non-augmented training loader created.")
print("Training images:", len(no_aug_train_dataset))
print("Classes:", len(no_aug_train_dataset.classes))
print("Random flips and rotations: disabled")
```

## Train ViT without augmentation

```python
import time
import torch
from torchvision.models import vit_b_16, ViT_B_16_Weights
from torch import nn

# Start with pretrained weights and a fresh classification head
vit_noaug_model = vit_b_16(weights=ViT_B_16_Weights.DEFAULT)
vit_noaug_model.heads.head = nn.Linear(
    vit_noaug_model.heads.head.in_features,
    len(no_aug_train_dataset.classes)
)
vit_noaug_model = vit_noaug_model.to(device)

noaug_criterion = nn.CrossEntropyLoss()
noaug_optimizer = torch.optim.AdamW(
    vit_noaug_model.parameters(),
    lr=0.0001,
    weight_decay=0.01
)

noaug_history = []
best_noaug_val_accuracy = 0.0

for epoch in range(5):
    start_time = time.time()

    # Training
    vit_noaug_model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in no_aug_train_loader:
        images, labels = images.to(device), labels.to(device)

        noaug_optimizer.zero_grad()
        outputs = vit_noaug_model(images)
        loss = noaug_criterion(outputs, labels)
        loss.backward()
        noaug_optimizer.step()

        running_loss += loss.item() * images.size(0)
        correct += (outputs.argmax(1) == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / total
    train_accuracy = correct / total

    # Validation
    vit_noaug_model.eval()
    running_val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in valid_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = vit_noaug_model(images)
            loss = noaug_criterion(outputs, labels)

            running_val_loss += loss.item() * images.size(0)
            val_correct += (outputs.argmax(1) == labels).sum().item()
            val_total += labels.size(0)

    val_loss = running_val_loss / val_total
    val_accuracy = val_correct / val_total

    noaug_history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "val_loss": val_loss,
        "val_accuracy": val_accuracy
    })

    if val_accuracy > best_noaug_val_accuracy:
        best_noaug_val_accuracy = val_accuracy
        torch.save(
            vit_noaug_model.state_dict(),
            "/content/best_vit_noaug_model.pth"
        )

    print(
        f"Epoch [{epoch+1}/5] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Time: {time.time() - start_time:.1f}s"
    )

print(f"\nBest no-augmentation validation accuracy: "
      f"{best_noaug_val_accuracy:.4f}")
print("Best checkpoint: /content/best_vit_noaug_model.pth")
```

## Implement attention rollout

```python
import torch
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
import types
from PIL import Image

from torchvision.models import vit_b_16
from torchvision import transforms
from torch.utils.data import DataLoader

# 1. Load the saved augmented ViT
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

vit_attention_model = vit_b_16(weights=None)
vit_attention_model.heads.head = torch.nn.Linear(
    vit_attention_model.heads.head.in_features, 27
)

checkpoint = torch.load(
    "/content/best_vit_model.pth",
    map_location=device,
    weights_only=True
)
vit_attention_model.load_state_dict(checkpoint)
vit_attention_model = vit_attention_model.to(device)
vit_attention_model.eval()

# 2. Capture attention weights from each Transformer block
encoder_blocks = vit_attention_model.encoder.layers

for block in encoder_blocks:
    original_forward = block.forward

    def forward_with_attention(self, x):
        residual = x
        normalized = self.ln_1(x)

        attention_output, attention_weights = self.self_attention(
            normalized, normalized, normalized,
            need_weights=True,
            average_attn_weights=False
        )
        self.attention_weights = attention_weights.detach()

        x = residual + self.dropout(attention_output)
        x = x + self.mlp(self.ln_2(x))
        return x

    block.forward = types.MethodType(forward_with_attention, block)

# 3. Get test predictions to find correct and incorrect examples
attention_loader = DataLoader(
    test_dataset, batch_size=32, shuffle=False, num_workers=2
)

all_preds, all_true = [], []

with torch.no_grad():
    for images, labels in attention_loader:
        outputs = vit_attention_model(images.to(device))
        all_preds.extend(outputs.argmax(dim=1).cpu().tolist())
        all_true.extend(labels.tolist())

correct_indices = [
    i for i, (p, y) in enumerate(zip(all_preds, all_true)) if p == y
]
incorrect_indices = [
    i for i, (p, y) in enumerate(zip(all_preds, all_true)) if p != y
]

print("Correct predictions:", len(correct_indices))
print("Incorrect predictions:", len(incorrect_indices))

# 4. Calculate attention rollout for a single image
def get_attention_rollout(image_tensor):
    with torch.no_grad():
        _ = vit_attention_model(image_tensor.unsqueeze(0).to(device))

    rollout = None

    for block in encoder_blocks:
        attention = block.attention_weights[0].mean(dim=0)
        identity = torch.eye(
            attention.shape[-1], device=attention.device
        )

        attention = attention + identity
        attention = attention / attention.sum(dim=-1, keepdim=True)

        if rollout is None:
            rollout = attention
        else:
            rollout = attention @ rollout

    # First token is CLS; remaining tokens represent image patches
    cls_attention = rollout[0, 1:]

    grid_size = int(cls_attention.numel() ** 0.5)
    heatmap = cls_attention.reshape(grid_size, grid_size)
    heatmap = heatmap.cpu().numpy()

    heatmap = (heatmap - heatmap.min()) / (
        heatmap.max() - heatmap.min() + 1e-8
    )

    return heatmap

# 5. Display a correct and an incorrect prediction
mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

chosen_indices = []
if correct_indices:
    chosen_indices.append(("Correct prediction", correct_indices[0]))
if incorrect_indices:
    chosen_indices.append(("Incorrect prediction", incorrect_indices[0]))

if not chosen_indices:
    print("No test examples were found.")
else:
    fig, axes = plt.subplots(
        len(chosen_indices), 3,
        figsize=(12, 4 * len(chosen_indices)),
        squeeze=False
    )

    for row, (description, index) in enumerate(chosen_indices):
        image_tensor, true_label = test_dataset[index]
        predicted_label = all_preds[index]

        # Original RGB image for visualization
        image = image_tensor * std + mean
        image = image.clamp(0, 1).permute(1, 2, 0).numpy()

        heatmap = get_attention_rollout(image_tensor)

        heatmap_resized = np.array(
            Image.fromarray(
                (heatmap * 255).astype(np.uint8)
            ).resize((224, 224), Image.Resampling.BILINEAR)
        ) / 255.0

        axes[row, 0].imshow(image)
        axes[row, 0].set_title(
            f"{description}\nTrue: {test_dataset.classes[true_label]}\n"
            f"Predicted: {test_dataset.classes[predicted_label]}"
        )

        axes[row, 1].imshow(heatmap_resized, cmap="jet")
        axes[row, 1].set_title("Attention rollout")

        axes[row, 2].imshow(image)
        axes[row, 2].imshow(
            heatmap_resized, cmap="jet", alpha=0.45
        )
        axes[row, 2].set_title("Attention overlay")

        for ax in axes[row]:
            ax.axis("off")

    plt.tight_layout()
    plt.show()
```
